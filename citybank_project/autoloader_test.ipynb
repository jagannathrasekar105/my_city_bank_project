```python
source_path="/Volumes/city_bank/source/fraud_watchlist/source_data/"
```

```python
dbutils.fs.ls(source_path)
```

```python
input_stream=spark.readStream.format("cloudFiles").option("cloudFiles.format","json").option("cloudFiles.schemaLocation","/Volumes/city_bank/source/fraud_watchlist/schema/").option("cloudFiles.inferColumnTypes","true").load(source_path)
```

```python
from pyspark.sql import functions as F
transformed_df=input_stream.select("*",F.col("_metadata.file_path").alias("file_path"),F.current_timestamp().alias("ingestion_timestamp"))

```

```python
streamming_query=(transformed_df.writeStream.format("delta").outputMode("append").option("checkpointLocation","/Volumes/city_bank/source/fraud_watchlist/checkpoint/").trigger(availableNow=True).toTable("city_bank.bronze.fraud_watchlist_batch_test"))
```
